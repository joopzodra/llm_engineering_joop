```python
from pathlib import Path
from tkinter import Tk, filedialog
import shutil


def select_and_copy_file(target_directory: str) -> Path | None:

    dest_dir = Path(target_directory).expanduser().resolve()

    if not dest_dir.exists():
        raise FileNotFoundError(f"Destination directory does not exist: {dest_dir}")
    if not dest_dir.is_dir():
        raise NotADirectoryError(f"Not a directory: {dest_dir}")

    # Hide the main Tk window and show only the file selection dialog.
    root = Tk()
    root.withdraw()
    root.attributes('-topmost', True)  # Bring dialog to front
    root.update()

    selected_file = filedialog.askopenfilename(title="Select a file to copy")
    root.destroy()

    if not selected_file:
        return None  # User canceled

    src = Path(selected_file).expanduser().resolve()
    if not src.is_file():
        raise FileNotFoundError(f"Selected path is not a file: {src}")

    destination = dest_dir / src.name
    shutil.copy2(src, destination)  # copy2 preserves metadata when possible
    return destination
```
